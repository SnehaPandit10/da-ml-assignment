```python
import pandas as pd
import numpy as np

path ='pokemon.csv'
df = pd.read_csv(path)
print(df)
```

```python
legendary_pokemon = df[df["is_legendary"] == 1]["name"]

print(legendary_pokemon)
```

```python
pokemon_name = input("Enter Pokemon name: ")

pokemon = df[df["name"].str.lower() == pokemon_name.lower()]

if not pokemon.empty:
    print(pokemon.to_string(index=False))
else:
    print("Pokemon not found")
```

```python
max_total = df["base_total"].max()

pokemon = df[df["base_total"] == max_total]
if len(pokemon) > 1:
    legendary = pokemon[pokemon["is_legendary"] == 1]

    if not legendary.empty:
        pokemon = legendary
if len(pokemon) > 1:
    gen3 = pokemon[pokemon["generation"] == 3]

    if not gen3.empty:
        pokemon = gen3

print(pokemon.to_string(index=False))
```
