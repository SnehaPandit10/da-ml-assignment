```python
import pandas as pd
import numpy as np
code=['c1','c2','c3','c4','c5']
courses = ['PGDBDA','PGDAI','PGDMC','PGDCSF','PGDGI']
lang=['python','python','java','c++','rprog']
df = pd.DataFrame({'code':code,'courses':courses,'languages':lang})
df
```

```python
from sklearn.preprocessing import LabelEncoder
l = LabelEncoder()
df['lang_encoded'] = l.fit_transform(df['languages'])
df
```

```python
data = ['PGDBDA','PGDAI','PGDMC','PGDCSF','PGDGI']
l.fit_transform(data)
l.classes_
e_data = l.transform(data)
```

```python
d = pd.DataFrame({})
d['course_encoded'] = e_data
d
```

```python
from sklearn.preprocessing import OneHotEncoder
o = OneHotEncoder(sparse_output=False)
o_c = o.fit_transform(df[['courses']])
encoded_df = pd.DataFrame(o_c, columns=o.get_feature_names_out(['courses']))
d = pd.concat([d,encoded_df],axis=1)
d
```
