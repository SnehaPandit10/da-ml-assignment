```python
import pandas as pd
import numpy as np

path = "car.csv"

df = pd.read_csv(path)
df.loc[df.groupby("Make")["MSRP"].idxmax()][["Make", "Model", "Year", "MSRP"]]
```

```python
convertibles = df[df["Vehicle Style"].str.lower() == "convertible"]

convertibles.loc[convertibles.groupby("Make")["MSRP"].idxmax()][
    ["Make", "Model", "Year", "Vehicle Style", "MSRP"]
]
```

```python
automatic_cars = df[df["Transmission Type"].str.upper() == "AUTOMATIC"]

automatic_cars
```

```python
premium_cars = df[
    df["Engine Fuel Type"].str.lower().str.contains("premium", na=False)
]

premium_cars[["Make", "Model", "Year", "Engine Fuel Type", "MSRP"]]
```

```python
df.groupby("Make")["MSRP"].mean().sort_values(ascending=False)
```
